```python
secret_code = "charlie-bravo-alpha"
secret_code = secret_code.replace("-", " ")
secret_code = secret_code.replace("charlie", "humm")
secret_code = secret_code.replace("alpha", "charlie")
secret_code = secret_code.replace("humm", "alpha")



print(secret_code.upper())
```

```python
secret_code = "charlie-bravo-alpha"
lst = secret_code.split("-")
lst = lst[::-1]
lst = [i.upper() for i in lst]
print(lst)
ans = " ".join(lst)
print(ans)
```
