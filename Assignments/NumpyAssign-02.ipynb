Q1.
A class has 4 students and 3 subjects. Their marks are given below. The examiner announces a subject-wise bonus of 5, 2 and 8 marks respectively. Add the bonus to every student in one line using broadcasting, then cap any mark above 100 back to 100 and report how many marks had to be capped.

marks = np.array([[88, 96, 95],
                  [72, 99, 91],
                  [98, 85, 97],
                  [60, 75, 80]])

```python
# Q1

marks = np.array([[88, 96, 95],
                  [72, 99, 91],
                  [98, 85, 97],
                  [60, 75, 80]])

bonus = np.array([5, 2, 8])

marks = marks + bonus

marks = np.minimum(marks, 100)

print(marks)

capped = np.sum(marks == 100)

print("Number of marks capped:", capped)
```

Q4.
Two arrays of size 4 are given. First join them with hstack and print the shape. Then, using the same two arrays, build a proper (4, 2) two-column table. In a comment, state why the first attempt did not produce two columns.

heights = np.array([172, 158, 181, 166])
weights = np.array([68, 52, 84, 59])

```python
# Q4

heights = np.array([172, 158, 181, 166])
weights = np.array([68, 52, 84, 59])

first = np.hstack((heights, weights))

print("First attempt:")
print(first)
print("Shape:", first.shape)

table = np.column_stack((heights, weights))

print("\nProper 2-column table:")
print(table)
print("Shape:", table.shape)

# np.hstack joins two 1-D arrays into one longer 1-D array,
# so its shape is (8,), not (4, 2).
```

Q8.
For the matrix below, scale every column into the 0 to 1 range using column-wise min-max normalization (x - col_min) / (col_max - col_min). Use broadcasting only — no loops. Verify that each column of the result has a minimum of 0 and a maximum of 1.

X = np.array([[ 50, 1200,  3],

Q9.
Award medals to the scores below using nested np.where: 'Gold' for 90 and above, 'Silver' for 75 and above, 'Bronze' for 60 and above, and 'None' otherwise. Then count how many students received each medal.

scores = np.array([94, 61, 77, 45, 90, 88, 59, 72])

```python
# Q9

scores = np.array([94, 61, 77, 45, 90, 88, 59, 72])

medals = np.where(
    scores >= 90,
    "Gold",
    np.where(
        scores >= 75,
        "Silver",
        np.where(
            scores >= 60,
            "Bronze",
            "None"
        )
    )
)

print(medals)

print("\nGold:", np.sum(medals == "Gold"))
print("Silver:", np.sum(medals == "Silver"))
print("Bronze:", np.sum(medals == "Bronze"))
print("None:", np.sum(medals == "None"))
```

m = np.array([[10, 80, 45],

Q11.
Three sensors each recorded 5 readings. Stack them so that each sensor becomes a row, and separately stack them so that each sensor becomes a column. Print both shapes and confirm that one is the transpose of the other.

s1 = np.array([12, 15, 11, 19, 14])
s2 = np.array([22, 18, 25, 21, 20])
s3 = np.array([31, 35, 29, 33, 30])

```python
# Q11

s1 = np.array([12, 15, 11, 19, 14])
s2 = np.array([22, 18, 25, 21, 20])
s3 = np.array([31, 35, 29, 33, 30])

rows = np.vstack((s1, s2, s3))
columns = np.column_stack((s1, s2, s3))

print("Sensors as rows:")
print(rows)
print("Shape:", rows.shape)

print("\nSensors as columns:")
print(columns)
print("Shape:", columns.shape)

print("\nTranspose check:")
print(np.array_equal(rows.T, columns))
```

Q13.
A and B are both 2x3 matrices. Compute A @ B.T and A.T @ B. Print the shape of each result and explain in a comment why both multiplications are legal while A @ B is not.

A = np.array([[1, 2, 3],
              [4, 5, 6]])
B = np.array([[7, 8, 9],
              [1, 0, 2]])

```python
# Q13

A = np.array([[1, 2, 3],
              [4, 5, 6]])

B = np.array([[7, 8, 9],
              [1, 0, 2]])

result1 = A @ B.T

print("A @ B.T:")
print(result1)
print("Shape:", result1.shape)

result2 = A.T @ B

print("\nA.T @ B:")
print(result2)
print("Shape:", result2.shape)

# Both are legal because:
# A is (2,3), B.T is (3,2)  -> result (2,2)
# A.T is (3,2), B is (2,3)  -> result (3,3)
# A @ B is not legal because (2,3) @ (2,3)
# has mismatched inner dimensions 3 and 2.
```

Q14.
Build a complete multi-step pipeline. Starting from the two arrays below: convert them into column vectors, join them into a single two-column table, attach a third column holding each row's total, then multiply the three-column table by w = [0.5, 0.2, 0.3] to obtain one score per row. Finally use np.where to mark each score as 1 if it is above the average score and 0 otherwise. Report the shape of every intermediate array.

hours = np.array([2, 8, 5, 10, 3, 7])
attendance = np.array([55, 92, 70, 98, 60, 85])

```python
# Q14

hours = np.array([2, 8, 5, 10, 3, 7])
attendance = np.array([55, 92, 70, 98, 60, 85])

# Step 1: Convert to column vectors
hours_col = hours.reshape(-1, 1)
attendance_col = attendance.reshape(-1, 1)

print("hours_col shape:", hours_col.shape)
print("attendance_col shape:", attendance_col.shape)

# Step 2: Join into 2-column table
table = np.hstack((hours_col, attendance_col))

print("\n2-column table:")
print(table)
print("Shape:", table.shape)

# Step 3: Add row total as third column
totals = table.sum(axis=1).reshape(-1, 1)

three_col = np.hstack((table, totals))

print("\nThree-column table:")
print(three_col)
print("Shape:", three_col.shape)

# Step 4: Weighted score
w = np.array([0.5, 0.2, 0.3])

scores = three_col @ w

print("\nScores:")
print(scores)
print("Shape:", scores.shape)

# Step 5: Average
average = scores.mean()

print("\nAverage score:", average)

# Step 6: Mark above average as 1, otherwise 0
result = np.where(scores > average, 1, 0)

print("\nFinal result:")
print(result)
print("Shape:", result.shape)
```

Q16.
Take the array below and create two 1D versions of it — one with flatten() and one with ravel(). Change the first element of each, then print the original matrix. Explain the difference you observe in a comment.

m = np.array([[4, 8, 15],
              [16, 23, 42]])

```python
# Q16

m = np.array([[4, 8, 15],
              [16, 23, 42]])

flat = m.flatten()
rav = m.ravel()

flat[0] = 999
rav[0] = 999

print("Original matrix:")
print(m)

print("\nflatten():")
print(flat)

print("\nravel():")
print(rav)

# flatten() creates a copy, so changing flat does not affect m.
# ravel() usually returns a view, so changing rav changes m.
```

Q17.
28 daily temperature readings are given. Reshape them into 4 weeks of 7 days, then: find each week's average, identify the hottest week, and count how many individual days crossed 35 degrees.

np.random.seed(7)
temps = np.random.randint(15, 41, 28)

```python
# Q17

np.random.seed(7)

temps = np.random.randint(15, 41, 28)

weeks = temps.reshape(4, 7)

print("4 weeks:")
print(weeks)

weekly_avg = weeks.mean(axis=1)

print("\nWeekly averages:")
print(weekly_avg)

hottest_week = weekly_avg.argmax()

print("\nHottest week index:", hottest_week)
print("Hottest week average:", weekly_avg[hottest_week])

days_above_35 = np.sum(temps > 35)

print("\nDays above 35°C:", days_above_35)
```

Q18.
Create two random 3x3 matrices A and B, compute A @ B and B @ A, and check whether the two results are equal. State your conclusion about matrix multiplication in a comment.

```python
# Q18

np.random.seed(10)

A = np.random.randint(1, 10, (3, 3))
B = np.random.randint(1, 10, (3, 3))

result_AB = A @ B
result_BA = B @ A

print("A:")
print(A)

print("\nB:")
print(B)

print("\nA @ B:")
print(result_AB)

print("\nB @ A:")
print(result_BA)

print("\nAre they equal?")
print(np.array_equal(result_AB, result_BA))

# Matrix multiplication is generally NOT commutative:
# A @ B != B @ A
```

Q20.
Start from arr = np.arange(1, 11). Create one variable using a plain slice arr[2:6] and another using arr[2:6].copy(). Change the first element of each to 999, then print the original array and explain in a comment which change survived and why.

```python
# Q20

arr = np.arange(1, 11)

view = arr[2:6]
copy = arr[2:6].copy()

view[0] = 999
copy[0] = 999

print("Original array:")
print(arr)

print("\nSlice view:")
print(view)

print("\nCopy:")
print(copy)

# arr[2:6] creates a view, so changing view changes arr.
# arr[2:6].copy() creates an independent copy,
# so changing copy does not affect arr.
```
