Q6. Given a = np.array([2, 4, 6]) and b = np.array([2.0, 4.0, 6.0]) - check both dtypes, then convert b to int with astype and verify the dtypes now match.

```python
a = np.array([2, 4, 6])
b = np.array([2.0, 4.0, 6.0])
print("a:", a)
print("a dtype:", a.dtype)
print("b:", b)
print("b dtype:", b.dtype)
b = b.astype(int)
print("After conversion:")
print("b:", b)
print("b dtype:", b.dtype)
```

Q15. Given the marks array below - (a) extract all marks greater than 60; (b) extract marks between 50 and 80 using two conditions with & (remember the parentheses!).

marks = np.array([56, 91, 44, 77, 62, 85, 39, 70])

```python
marks = np.array([56, 91, 44, 77, 62, 85, 39, 70])
# (a) Greater than 60
greater_60 = marks[marks > 60]
# (b) Between 50 and 80
between_50_80 = marks[(marks > 50) & (marks < 80)]
print("Marks greater than 60:")
print(greater_60)
print("\nMarks between 50 and 80:")
print(between_50_80)
```
