Part A — tips.csv (the restaurant)

A1. The owner wants to know how generous each customer was, not just how much they left.

Make a new column called tip_percent — the tip as a percentage of the bill, rounded to 1 decimal place. Then show only total_bill, tip and tip_percent for the first 5 rows.

```python
df['tip_percent'] = (df['tip'] / df['total_bill'] * 100).round(1)
print(df[['total_bill', 'tip', 'tip_percent']].head())
```

A4. Two counts for the owner:

How many bills were not at Dinner? Use ~.
How many bills had a party of 3 or more and a tip above 4?

```python
print((~(df['time'] == 'Dinner')).sum())
print(((df['size'] >= 3) & (df['tip'] > 4)).sum())
```

A5. The owner wants to flag the generous customers.

Make a column called service. Everyone starts as 'normal'. Any bill where tip_percent is more than 20 becomes 'generous'.

Then count how many of each. (This uses the column you made in A1.)

```python
df['service'] = 'normal'
df.loc[df['tip_percent'] > 20, 'service'] = 'generous'
print(df['service'].value_counts())
```

Part B — Titanic-Dataset.csv

⚠️ Remember: the column names here start with capital letters — Survived, Pclass, Sex, Age, SibSp, Parch, Fare, Cabin, Embarked.

B1. How many passengers boarded from each port? Use the Embarked column.

Then add the three numbers up in a comment. Do they come to 891? ⭐

```python
print(t['Embarked'].value_counts())
```

B2. Two counts:

How many women were travelling in third class?
How many of those women survived?

```python
women_third = (t['Sex'] == 'female') & (t['Pclass'] == 3)
print(women_third.sum())
print((women_third & (t['Survived'] == 1)).sum())
```

B3. Some people paid an enormous amount for their ticket.

Show only Name, Age and Fare for every passenger who paid more than 200. How many are there?

```python
expensive = t.loc[t['Fare'] > 200, ['Name', 'Age', 'Fare']]
print(expensive)
print(len(expensive))
```

B4. Out of all the passengers in first class, what percentage survived?

Now do the same for third class. Put the two numbers next to each other in a comment.

```python
first_class = t[t['Pclass'] == 1]
third_class = t[t['Pclass'] == 3]
first_percentage = first_class['Survived'].mean() * 100
third_percentage = third_class['Survived'].mean() * 100
print(first_percentage)
print(third_percentage)
```
