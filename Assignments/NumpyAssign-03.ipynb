```python
marks = np.array([[72, 85, 60, 91],
                  [45, 30, 98, 55],
                  [88, 66, 74, 100],
                  [95, 92, 55, 40]])

grace = np.array([5, 0, 8, 3])
weightage = np.array([1.2, 1.0, 1.1, 1.0])
```

```python
# Q10

# 1. Apply subject-wise grace, maximum 100
grace_marks = np.minimum(marks + grace, 100)
print(grace_marks)

# 2. Multiply each subject by its weightage
weighted = np.round(marks * weightage, 1)
print(weighted)

# 3. Difference from subject-wise class average
subject_mean = np.mean(marks, axis=0)
difference = marks - subject_mean
print(difference)

# 4. Give row-wise bonus
bonus = np.array([5, 0, 2, 0]).reshape(-1, 1)
bonus_result = marks + bonus
print(bonus_result)

# 5. Count entries below subject average
print(np.sum(difference < 0))

# 6. Standardize every subject
subject_std = np.std(marks, axis=0)
standardized = np.round(
    (marks - subject_mean) / subject_std,
    2
)

print(standardized)
```

Print the index positions of the Short students along with their attendance values.
In the marks table, find the row and column position of every mark below 50.

```python
marks = np.array([[70, 65, 80],
                  [55, 90, 32],
                  [88, 76, 91],
                  [40, 52, 45],
                  [60, 58, 66]])

attendance = np.array([88, -1, 105, 76, 45, -1, 92, 120, 67, 30])
```

```python
# Q14

# 1. Repair attendance

attendance_clean = attendance.copy()

# Values above 100 become 100
attendance_clean = np.minimum(attendance_clean, 100)

# Average of genuine entries
average = np.mean(attendance_clean[attendance_clean != -1])

# Replace -1 with average
attendance_clean = np.where(
    attendance_clean == -1,
    average,
    attendance_clean
)

print(attendance_clean)


# 2. Label attendance

labels = np.where(
    attendance_clean >= 85,
    "Regular",
    np.where(
        attendance_clean >= 60,
        "Average",
        "Short"
    )
)

print(labels)


# 3. Index and attendance of Short students

short_positions = np.where(attendance_clean < 60)[0]

print(short_positions)
print(attendance_clean[short_positions])


# 4. Row and column positions of marks below 50

rows, cols = np.where(marks < 50)

print("rows:", rows)
print("cols:", cols)


# 5. Actual marks at those positions

print(marks[rows, cols])


# 6. Replace every mark below 35 with 35
count_changed = np.sum(marks < 35)

marks[marks < 35] = 35

print(count_changed)
print(marks)


# 7. Student with highest number of marks below 60

below_60 = np.sum(marks < 60, axis=1)

print(np.argmax(below_60))
```

Q15. The final weighted result
Tasks

Compute the weighted final score of every student using w.
Print the index of the student with the highest weighted score.
Using a multiplication rather than sum, produce the subject-wise total marks of the whole class.
The board has proposed two weightage schemes, held in W. Compute both final scores of all four students in one multiplication.
From that result, print how many students score higher under the second scheme.
Print the difference between the two schemes for every student, rounded to 2 decimals.
Attach the better of the two scores as a fourth column to M.
Expected output — task 3

[273. 289. 309.]
Expected output — task 4

[[72.5 72. ]
 [72.3 67.1]
 [85.6 86.5]
 [61.8 61.4]]
Expected output — task 6

[-0.5 -5.2  0.9 -0.4]

```python
M = np.array([[70, 65, 80],
              [55, 90, 72],
              [88, 76, 91],
              [60, 58, 66]])

w = np.array([0.3, 0.3, 0.4])

W = np.array([[0.3, 0.5],
              [0.3, 0.2],
              [0.4, 0.3]])
```

```python
# Q15

# 1. Weighted final score
final_scores = M @ w

print(final_scores)


# 2. Index of student with highest weighted score
print(np.argmax(final_scores))


# 3. Subject-wise total using multiplication
ones = np.ones(M.shape[0])

subject_totals = ones @ M

print(subject_totals)


# 4. Both weighting schemes in one multiplication
both_scores = M @ W

print(both_scores)


# 5. Number of students scoring higher under second scheme
count = np.sum(both_scores[:, 1] > both_scores[:, 0])

print(count)


# 6. Difference between second and first scheme
difference = np.round(
    both_scores[:, 1] - both_scores[:, 0],
    2
)

print(difference)


# 7. Attach better score as fourth column
better_score = np.max(both_scores, axis=1).reshape(-1, 1)

result = np.hstack((M, better_score))

print(result)
```
