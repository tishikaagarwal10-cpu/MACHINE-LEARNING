Question 1: The Strict Username Validator
Task: Write a function that takes a username string as input and validates it. To be valid, the username must be at least 5 characters long after removing any leading/trailing spaces, and it must contain only letters and numbers (no special characters like @, _, or spaces inside). If it is valid, return the clean, strictly lowercase version of the username. If it is invalid, return the string "Invalid".

Input 1:

" CodeMaster99 "
Input 2:

" Python_Dev "
Expected Output 1:

"codemaster99"
Expected Output 2:

"Invalid"


```python
def validate_username(username):
    username = username.strip()

    if len(username) < 5:
        return "Invalid"

    if not username.isalnum():
        return "Invalid"

    return username.lower()


username1 = "  CodeMaster99  "
username2 = "  Python_Dev  "

print(validate_username(username1))
print(validate_username(username2))
```

Question 2: The Spy Code Reverser
Task: You intercept a secret code where words are separated by hyphens (-). To decode it, you must reverse the order of the words, convert every letter to UPPERCASE, and replace the hyphens with spaces.

secret_code = "charlie-bravo-alpha"

"ALPHA BRAVO CHARLIE"

Question 5: The Palindrome Sentence Checker
Task: Determine if a given sentence is a palindrome. To do this correctly using only string methods, remove all spaces and non-alphanumeric characters, convert everything to lowercase, and check if it reads the same forwards and backwards.

Input 1:

sentence = "Was it a car or a cat I saw?"
Expected Output 1:

True
Input 2:

sentence = "Python is fun"
Expected Output 2:

False

```python
def is_palindrome(sentence):
    clean = ""

    for ch in sentence:
        if ch.isalnum():
            clean += ch.lower()

    return clean == clean[::-1]


sentence1 = "Was it a car or a cat I saw?"
sentence2 = "Python is fun"

print(is_palindrome(sentence1))
print(is_palindrome(sentence2))
```

Question 7: The File Extension Finder
Task: You are given a single comma-separated string containing various filenames. Extract just the file extensions (without the dot), ensure there are no duplicates, and return them as a list of strings strictly in UPPERCASE.

Input:

filenames = "report.pdf, image.JPG, data.csv, notes.txt, photo.jpg"
Expected Output:

['PDF', 'JPG', 'CSV', 'TXT']

```python
filenames = "report.pdf, image.JPG, data.csv, notes.txt, photo.jpg"
files = filenames.split(",")
extensions = []
for file in files:
    file = file.strip()
    ext = file.split(".")[-1].upper()

    if ext not in extensions:
        extensions.append(ext)

print(extensions)
```

Question 8: The Domain Name Slicer
Task: Given an email address string, extract just the company/domain name (the part immediately after the @ symbol and before the dot suffix like .com or .edu). Convert the extracted domain to Title Case. If the string does not contain an @, return "Invalid".

Input 1:

email = "developer_99@microsoft.com"
Expected Output 1:

"Microsoft"
Input 2:

email = "admin_at_website.org"
Expected Output 2:

"Invalid"

```python
def get_domain(email):
    if "@" not in email:
        return "Invalid"

    domain = email.split("@")[1]
    company = domain.split(".")[0]

    return company.title()


email1 = "developer_99@microsoft.com"
email2 = "admin_at_website.org"

print(get_domain(email1))
print(get_domain(email2))
```

Question 9: The Anagram Detective
Task: Write a function to check if two given strings are anagrams of each other (they contain the exact same letters in the exact same frequencies, just rearranged). Ignore any spaces and treat uppercase and lowercase letters as the same. Solve it using string methods and basic built-in functions (like sorted()).

Input:

str1 = "Dormitory"
str2 = "Dirty room"
Expected Output:

True

```python
def is_anagram(str1, str2):
    str1 = str1.replace(" ", "").lower()
    str2 = str2.replace(" ", "").lower()

    return sorted(str1) == sorted(str2)


str1 = "Dormitory"
str2 = "Dirty room"

print(is_anagram(str1, str2))
```

Question 10: The Messy Phone Number Formatter
Task: You are building a CRM system and user-submitted phone numbers are a mess. They contain spaces, parentheses, periods, and hyphens. Write a function that takes a raw phone number string and cleans it by removing all non-digit characters.

Once cleaned, validate the number:

If the cleaned string contains exactly 10 digits, format and return it as "(XXX) XXX-XXXX".
If it contains anything other than 10 digits, return "Invalid Number".
Input 1:

raw_num = " +1 (555) 123.4567 "
Expected Output 1:

"Invalid Number"
(Because after removing symbols, it has 11 digits due to the '1')

Input 2:

raw_num = "555-987-6543"
Expected Output 2:

"(555) 987-6543"

```python
def format_phone(raw_num):
    digits = ""

    for ch in raw_num:
        if ch.isdigit():
            digits += ch

    if len(digits) != 10:
        return "Invalid Number"

    return "(" + digits[:3] + ") " + digits[3:6] + "-" + digits[6:]


raw_num1 = " +1 (555) 123.4567 "
raw_num2 = "555-987-6543"

print(format_phone(raw_num1))
print(format_phone(raw_num2))
```

Question 11: The Product SKU Inspector
Task: An inventory system requires product codes (SKUs) to strictly match this format after being cleaned up: exactly 3 uppercase letters, a single hyphen, and exactly 4 digits (e.g., "ABC-1234").

Write a function that receives a messy SKU string. First, clean it by stripping leading/trailing spaces and converting it to uppercase. Then, validate it using string methods (isalpha(), isdigit(), etc.) without using the re module.

Return "Valid: [Cleaned_SKU]" if it passes, or "Invalid SKU" if it fails any formatting rule.

Input 1:

sku = " xYz-9876 "
Expected Output 1:

"Valid: XYZ-9876"
Input 2:

sku = "AB-12345"
Expected Output 2:

"Invalid SKU"
(Because it has 2 letters and 5 digits)

Input 3:

sku = "A@C-1234"
Expected Output 3:

"Invalid SKU"
(Because it contains a special character)

```python
def check_sku(sku):
    sku = sku.strip().upper()

    if len(sku) != 8:
        return "Invalid SKU"

    if sku[3] != "-":
        return "Invalid SKU"

    if not sku[:3].isalpha():
        return "Invalid SKU"

    if not sku[4:].isdigit():
        return "Invalid SKU"

    return "Valid: " + sku


print(check_sku(" xYz-9876 "))
print(check_sku("AB-12345"))
print(check_sku("A@C-1234"))
```

Question 1: Selective Word Frequency Counter
Task: Write a function that counts the frequency of words in a sentence, but ignores any words present in a provided stop_words list. Store the result in a dictionary. All comparisons should be case-insensitive.

Input:

text = "The quick brown fox jumps over the lazy dog"
stop_words = ["the", "over", "is", "a"]
Expected Output:

{'quick': 1, 'brown': 1, 'fox': 1, 'jumps': 1, 'lazy': 1, 'dog': 1}

```python
def word_frequency(text, stop_words):
    words = text.lower().split()
    stop_words = [word.lower() for word in stop_words]

    result = {}

    for word in words:
        if word not in stop_words:
            if word in result:
                result[word] += 1
            else:
                result[word] = 1

    return result


text = "The quick brown fox jumps over the lazy dog"
stop_words = ["the", "over", "is", "a"]

print(word_frequency(text, stop_words))
```

Question 2: The Inventory Merger (Summing Overlapping Keys)
Task: You are managing a store with two branches. Each branch sends you its inventory as a dictionary. Write a code to merge these two dictionaries into a single master inventory. If an item exists in both branches, their quantities should be added together. If it exists in only one, just copy it over.

Input:

branch_A = {"laptop": 10, "mouse": 50, "keyboard": 20}
branch_B = {"mouse": 30, "monitor": 15, "laptop": 5}
Expected Output:

{'laptop': 15, 'mouse': 80, 'keyboard': 20, 'monitor': 15}

```python
branch_A = {"laptop": 10, "mouse": 50, "keyboard": 20}
branch_B = {"mouse": 30, "monitor": 15, "laptop": 5}

inventory = branch_A.copy()

for item, quantity in branch_B.items():
    if item in inventory:
        inventory[item] += quantity
    else:
        inventory[item] = quantity

print(inventory)
```

Question 3: The Dictionary Inverter (Grouping by Value)
Task: You have a dictionary where the keys are student names and the values are the grades they received. You need to "invert" this dictionary so that the grades become the keys, and the values become a list of student names who received that grade.

Input:

grades = {"Ravi": "A", "Neha": "B", "Aman": "A", "Pooja": "C", "Rahul": "B"}
Expected Output:

{'A': ['Ravi', 'Aman'], 'B': ['Neha', 'Rahul'], 'C': ['Pooja']}

```python
grades = {
    "Ravi": "A",
    "Neha": "B",
    "Aman": "A",
    "Pooja": "C",
    "Rahul": "B"
}

result = {}

for student, grade in grades.items():

    if grade not in result:
        result[grade] = []

    result[grade].append(student)

print(result)
```

Question 5: The Safe API Extractor
Task: You receive a list of user profiles from an API. Some profiles are incomplete and missing the "address" or "city" key. Write a function that processes this list and creates a simple dictionary mapping the user's "id" to their "city". If the city or address is missing, assign the value "Unknown". Do not use try-except, and do not let the program crash with a KeyError.

Input:

users = [
    {"id": 101, "name": "Aman", "profile": {"address": {"city": "Delhi"}}},
    {"id": 102, "name": "Neha", "profile": {}}, # Missing address entirely
    {"id": 103, "name": "Rahul", "profile": {"address": {"zip": "400001"}}} # Has address, but missing city
]
Expected Output:

{101: 'Delhi', 102: 'Unknown', 103: 'Unknown'}

```python
users = [
    {
        "id": 101,
        "name": "Aman",
        "profile": {
            "address": {
                "city": "Delhi"
            }
        }
    },
    {
        "id": 102,
        "name": "Neha",
        "profile": {}
    },
    {
        "id": 103,
        "name": "Rahul",
        "profile": {
            "address": {
                "zip": "400001"
            }
        }
    }
]

result = {}

for user in users:

    user_id = user["id"]

    profile = user.get("profile", {})
    address = profile.get("address", {})
    city = address.get("city", "Unknown")

    result[user_id] = city

print(result)
```
