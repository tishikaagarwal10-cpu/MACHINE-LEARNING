```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

titan = pd.read_csv("/Users/tishikaagarwal/Desktop/Machine Learning/Data/Titanic-Dataset.csv")
mpg = pd.read_csv("/Users/tishikaagarwal/Desktop/Machine Learning/Data/mpg.csv")

print(titan.shape, mpg.shape)
```

```python
titan.head(1)
```

```python
#THE MATPLOTLIB SKELETON

plt.figure(figsize=(7,4))
sns.histplot(data=titan, x='Fare', y='Age')
plt.title('fare')
plt.grid()
plt.show()
```

UNIVARIATE - ONE COLUMN AT A TIME

```python
# distribution of fare

plt.figure(figsize=(7,4))
sns.histplot(data=titan, x='Fare', bins=20)
plt.title('fare')
plt.grid()
plt.show()
```

```python
# distribution of age

plt.figure(figsize=(7,4))
sns.histplot(data=titan, x='Age', bins=20)
plt.title('Age Distrubution')
plt.show()
```

```python
#countplot - how many in each category
#how many survived and how many died

plt.figure(figsize=(7,4))
sns.countplot(data=titan, x='Survived')
plt.title('Survived')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.histplot(data=mpg, x='mpg', bins=25)
plt.title('mpg Data')
plt.show()
```

```python
#Q2

plt.figure(figsize=(7,4))
sns.countplot(data=mpg, x='origin')
plt.title('Origin data')
plt.show()
```
