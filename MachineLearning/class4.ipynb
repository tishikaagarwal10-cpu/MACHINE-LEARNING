```python
import numpy as np
```

```python
marks = np.array([35, 68, 42, 91, 27, 55])
print(np.where(marks>50, "pass" , "fail"))
# print(np.where(marks>50))
```

```python
marks = np.array([92, 61, 45, 78, 33, 55])
condi = np.where(marks>=75, 'A', 
        np.where(marks>=50 , 'B' , 'C'))
print(marks)
print(condi)
condi2 = np.where(marks>=50, 'B', 
        np.where(marks>=75 , 'A' , 'C'))
print(condi2)
```

```python
marks = np.array([92, 61, 45, 78, 33, 55])
idx = np.where(marks > 50)[0]
print("Raw return:" , idx)
idx2 = np.where(marks > 50)
print(idx2)
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print("a * b   :" , a * b)
print("np.dot  :" , np.dot(a,b))
print("a@b     :" , a @ b)
```

```python
a = np.array([[1, 2, 3], [7,8,9]])
b = np.array([[4, 5, 6], [10, 11, 12]])
print(a)
print(a.T)
print(a@a.T)
# print(a@b)
```
