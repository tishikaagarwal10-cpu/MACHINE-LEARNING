```python
#BASIC STATISTICS

arr = np.arange(6).reshape(2,3)
print(arr)
print("min", arr.min())
print("row minimum" , arr.min(axis=1))
print("column minimum" , arr.min(axis=0))

print("min", arr.max())
print("row minimum" , arr.max(axis=1))
print("column minimum" , arr.max(axis=0))
```

```python
arr = np.arange(6).reshape(2,3)
print(arr)
print("sum of the arr" , arr.sum())
print("sum of the row" , arr.sum(axis=1))
print("sum of the col" , arr.sum(axis=0))
```

```python
arr = np.arange(6).reshape(2,3)
print(arr)
print(np.median(arr))
print(np.median(arr, axis=0)) #col
print(np.median(arr, axis=1)) #row
```
