```python
titan = titan.drop(columns=["Cabin", "Name", "Ticket", "PassengerId"])
```

```python
titan["Age"] = titan["Age"].fillna(titan["Age"].median())
titan.dropna(subset=["Embarked"], inplace=True)
```

```python
titan.info()
```

```python
x = titan.drop(columns=["Survived"])
y = titan['Survived']
print(x.head(1))
print(y.head(1))
```

```python
from sklearn.model_selection import train_test_split
```

```python
#split

x_train, x_test, y_train , y_test = train_test_split(x,y,test_size=0.2, random_state=69)
```

`random_state` — same rows for everyone. `stratify` — same survival ratio in both halves.

```python
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier()
model.fit(x_train, y_train)
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

```

```python
cat_cols = ["Sex" , "Embarked"]
```

```python
ct = ColumnTransformer([('cat' , OneHotEncoder(sparse_output=False), cat_cols)] ,remainder='passthrough')

x_train_final = ct.fit_transform(x_train)
x_test_final = ct.transform(x_test)
```

```python
print(x_train.shape)
print(x_test.shape)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
model = KNeighborsClassifier()
model.fit(x_train_final, y_train)
y_predict = model.predict(x_test_final)
print(accuracy_score(y_test, y_predict))
```
