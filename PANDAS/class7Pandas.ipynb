
2️⃣ iloc — selecting by position
You did this in NumPy. Here we use it to pick rows.

Flow: one row → many rows → rows and columns together

```python
df.head(3)
```

```python
# iloc — one row, then a slice of rows
# df.iloc[4] for single row
# df.iloc[6:10:2]
# 6,7,8,10
# df.iloc[[6,7,8,10]] # fancy indexing 
df.iloc[(df['bill_per_person']>4).values]

# df.iloc[2,[1,3,5]]
```

```python
# iloc — rows and columns by position
```

```python
df.iloc[3]
print(df.iloc[1:10:2, [1,6]])

print(df.iloc[(df['bill_per_person']>4).values])
```
