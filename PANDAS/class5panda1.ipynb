```python
# pd.Series() - 1D data with an index

total_bill = [4700, 8000, 1200, 500]
s = pd.Series(total_bill, index=['mon', 'tue', 'wed', 'fri'])
print(s)
print(type(s))
```

```python
# pd.DataFrame() - 1D data with an index
restaurant_data = {
    'Table_No': [12, 14, 15],
    'Bill_Amount': [1250, 890, 2400],
    'Waiter_Name': ['Amit', 'Priya' , 'Rahul']
}

df = pd.DataFrame(restaurant_data)

print(df)
print(type(df))

print(df['Bill_Amount'])
print(type(df['Bill_Amount']))
```
