```python
tips.groupby("day").agg(
    Bill_count = ("total_bill" , 'size'),
    avg_bill = ("total_bill" , 'mean'),
    avg_tip = ("tip" , 'mean')
)
```

```python
tips.groupby("time").agg(
    Bill_count = ("total_bill" , 'size'),
    avg_bill = ("total_bill" , 'mean'),
    avg_party = ("size" , 'mean')
)
```

```python
summary = tips.groupby("day").agg(
    Bill_count = ("total_bill" , 'size'),
    avg_bill = ("total_bill" , 'mean'),
    avg_party = ("size" , 'mean')
).round(2)
summary
```

```python
summm = tips.groupby("smoker").agg(
    num_bill = ("total_bill" , 'size'),
    avg_bill = ("total_bill" , 'mean'),
    avg_tip = ("tip" , 'mean')
).round(2)
summm
```

```python
summm.reset_index().to_csv("smoker_report.csv", index=False)
report = pd.read_csv("smoker_report.csv")
report.head()
```
